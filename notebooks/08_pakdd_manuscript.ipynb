# 08 · PAKDD 2027 draft: every number in the manuscript

`paper/pakdd/main.tex` frames the study as a seed-0 2×2 comparison (pairwise vs. set-aware scorer × clean vs.
intervention training) with adapted ComEM and direct LLM selectors as supplementary baselines. Its numbers are
LaTeX macros and table rows written by `paper/pakdd/generate_results.py` from the locked runs. This notebook runs the same
generator, shows every macro and table, checks that the committed `paper/pakdd/generated/*.tex` are current, and adds
per-cohort counts for the systems in Table 5.

The archived scores under `runs/preds/` are included in this artifact package.

```python
import sys, json
sys.path.insert(0, '.')
from _nb import *
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.width', 200)
import importlib.util
spec = importlib.util.spec_from_file_location('pakdd_gen', REPO / 'paper/pakdd/generate_results.py')
gen = importlib.util.module_from_spec(spec); spec.loader.exec_module(gen)
tables = gen.build_tables(REPO / 'runs')
stale = [n for n, body in tables.items() if (REPO / 'paper/pakdd/generated' / n).read_text() != gen.GENERATED + body]
print('generated/*.tex up to date' if not stale else f'STALE, run `make -C paper/pakdd numbers`: {stale}')
```

## Macros used in the text

```python
import re
macros = re.findall(r'\\newcommand\{\\(\w+)\}\{(.*)\}', tables['results.tex'])
clean = lambda v: re.sub(r'\\ensuremath\{(.*)\}', r'\1', v).replace('\\%', '%').replace('\\,', ' ').replace('\\pm', '±').replace('--', '–')
md(pd.DataFrame([(k, clean(v)) for k, v in macros], columns=['macro', 'value']).set_index('macro'))
```

## Manuscript tables and supplemental transfer summary

```python
def rows(name, columns):
    body = [l for l in tables[name].splitlines() if l.strip() and not l.startswith(('\\midrule', '\\bottomrule'))]
    cells = [[clean(c.strip().rstrip('\\').strip()).replace('$0^\\dagger$', '0†') for c in l.split('&')] for l in body]
    return pd.DataFrame(cells, columns=columns)
specs = {
    'Table 1 · seed-0 2×2 on the locked test (%)': ('main_rows.tex', ['scorer', 'interv.', 'forced top-1', 'recall [95% CI]', 'precision', 'N-FMR', 'defer', 'correct NONE', 'false NONE']),
    'Table 2 · recall by dataset (%; † = scope forced to defer)': ('dataset_rows.tex', ['scorer', 'interv.', 'Abt-Buy', 'DBLP-Scholar', 'Walmart-Amazon', 'WDC Products']),
    'Table 3 · negative cohorts, main model': ('negative_rows.tex', ['dataset', 'cohort', 'false links / pools', 'affected / families', 'upper bound']),
    'Table 4 · perturbation diagnostics (%)': ('robustness_rows.tex', ['scorer', 'training', 'selected after removal', 'flipped after text edit']),
    'Table 5 · selector operating points on the locked test (%)': ('comem_rows.tex', ['system', 'prompt or policy', 'recall', 'precision', 'N-FMR', 'false NONE', 'defer']),
    'Supplemental · transfer to an unseen dataset (%)': ('transfer_rows.tex', ['held-out dataset', 'pairwise recall', 'pairwise precision', 'set recall', 'set precision']),
}
from IPython.display import Markdown, display
for title, (name, cols) in specs.items():
    display(Markdown(f'**{title}**')); md(rows(name, cols).set_index(cols[0]))
```

## Ranking vs. linking: forced top-1 by dataset

Forced top-1 ignores NONE and the thresholds: is an acceptable candidate ranked first? All four arms rank well; what differs
is how much each policy may link (Table 1). WDC is ranked well yet every scope defers, because no point of the frozen grid
qualifies on its calibration fit.

```python
ev = evaluation('runs/' + gen.PRIMARY_RUN)
top, by_ds = gen.forced_top_one(REPO / 'runs', int(ev['results'][gen.SET_ARM]['n_matchable']))
md(pd.DataFrame({ARM_LABEL.get(a, a): {'pooled': f3(top[a]), **{DS_LABEL[d]: f3(v) for d, v in by_ds[a].items()},
                  'policy recall': f3(ev['results'][a]['match_recall'])} for a, *_ in gen.ARMS}).T)
```

## External selectors and the main model on the same pools

The external selectors emit actions without a fitted deferral policy. ComEM shortlists four records with FLAN-T5-XL before
selection by GPT-4o mini. GPT-OSS and Gemma select directly from the supplied pool. Each row uses the same scorable pools
within its cohort. The main model uses its frozen 2% policy.

```python
ce = evaluation('runs/20261107T170250Z_comem_compare_test_60fe34')
rows_ = []
for result, arm, label in [
    (ev, 'judge_120b_rich', 'GPT-OSS-120B, rich'),
    (ev, 'judge_12b_minimal_json', 'Gemma 4 12B, minimal JSON'),
    (ev, 'judge_12b_rich', 'Gemma 4 12B, rich'),
    (ev, 'judge_12b_symbols', 'Gemma 4 12B, symbols'),
    (ce, 'comem', 'adapted ComEM'),
    (ce, 'set_joint_interv_seed0', 'main (frozen 2% policy)'),
]:
    for cohort, m in result['by_cohort'][arm]['__all__'].items():
        rows_.append({'system': label, 'cohort': cohort, 'false links': f"{int(m['SM'])} / {int(m['n_certified_nomatch'])}",
                      'correct NONE': f"{int(m['CR'])}", 'recall': f3(m['match_recall']) if m['n_matchable'] else '–'})
md(pd.DataFrame(rows_).set_index('system'))
s = load('runs/20261107T160617Z_comem_frozen_test/comparison_summary.json')
t4 = s['top4_acceptable_retention']['__all__']
print(f"acceptable candidate kept in ComEM's top-4 shortlist: {t4['retained_pools']} / {t4['matchable_pools']} matchable pools")
```
